$\begin{equation}
\nabla\cdot \vec{u} = 0,
\end{equation}$

\frac{\partial \vec{u}}{\partial t}+\vec{u}\cdot\nabla\vec{u} = -\nabla p + PrRa_{T,q}\underline{T}\vec{e}_z+Pr\nabla^2\vec{u},
\end{equation}$

$\begin{equation}
\frac{\partial \underline{T}}{\partial t}+\vec{u}\cdot\nabla\underline{T} = -\nabla^2 \underline{T}.
\end{equation}$

Computational domain:
- Depth $h=L_z=1$ with the number of points $N_z=64$
- Length of x-dimension $L_x=4$ with the number of points $N_x=256$


Rayleigh number and Prandtl number:
$
\begin{equation}
Ra_{T,q} := \frac{\alpha g |q| h^4}{\nu\kappa_T},\quad Pr := \frac{\nu}{\kappa_T}
\end{equation}
$

$
\begin{equation}
\Rightarrow \quad \kappa_T := (\alpha g |q| h^4)^{1/2} (Ra_{T,q}Pr)^{-1/2},\quad \nu := (\alpha g |q| h^4)^{1/2} \left(\frac{Ra_{T,q}}{Pr}\right)^{-1/2}
\end{equation}
$
